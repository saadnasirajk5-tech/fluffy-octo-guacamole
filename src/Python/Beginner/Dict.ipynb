```python
# Dictionary 
"""More direct way of associating an identifier with information.
   The key is the word you look up and value is the information associated 
   with that word.
"""
# Create a dictionary that maps task names to their owners.
owners = {
    "learn Python": "Saadi",  # Store the owner of the Python task.
    "practice HTTP": "Ali",  # Store the owner of the HTTP task.
    "write tests": "Saadi",  # A value may appear more than once.
}
python_owner = owners["learn Python"] 
owners["design API"] = "Ali" 
owners["write tests"] = "Ali" 
api_owner = owners.get("design API", "Unassigned")
# Ask for a key that does not exist, using a fallback value.
unknown_owner = owners.get("deploy application", "Unassigned") 
has_tests_task = "write tests" in owners 

print("Owners:", owners)
print("Python task owner:", python_owner)

print("API task owner:", api_owner)
print("Unknown task owner:", unknown_owner)
print("Tests task exists:", has_tests_task)

```
