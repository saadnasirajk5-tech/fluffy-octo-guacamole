```python
# Import Counter from Python's standard collections module.
from collections import Counter
# Store sample HTTP response status codes.
status_codes = [200, 200, 404, 200, 500, 404, 200, 503, 500]
# Count how many times each status code appears.
status_counts = Counter(status_codes)

# Print all status counts.
print(status_counts)
# Retrieve the number of successful 200 responses.
print("200 responses:", status_counts[200])
# Retrieve the number of 404 responses.
print("404 responses:", status_counts[404])
# Return the three most frequently occurring status codes.
print("Top three:", status_counts.most_common(2))
```

```python
# Count words in a sentence 
# Import Counter for frequency counting.
from collections import Counter
# Store a sample sentence as a string.
sentence = "Python makes APIs useful and Python makes automation useful"
# Split the sentence into individual words.
words = sentence.lower().split()
# Count the occurrences of each word.
word_counts = Counter(words)

# Print the resulting frequency counter.
print(word_counts)
# Display the five most frequent words.
print(word_counts.most_common(5))
# Check how often the word "python" appears.
print("Python frequency:", word_counts["python"])
```

```python
# Import the two data structures needed for nested frequency counting.
from collections import Counter, defaultdict
# Store a small sample of API requests as dictionaries.
requests = [
    {"path": "/users", "status": 200},  # Successful user request.
    {"path": "/users", "status": 404},  # Missing user.
    {"path": "/users", "status": 200},  # Another successful request.
    {"path": "/tasks", "status": 500},  # Server error.
    {"path": "/tasks", "status": 200},  # Successful task request.
    {"path": "/tasks", "status": 200},  # Another successful request.
]
# Create a Counter automatically for each endpoint encountered.
status_by_path = defaultdict(Counter)
# Iterate over each request record.
for request in requests:
    # Read the endpoint path from the current record.
    path = request["path"]
    # Read the HTTP status code from the current record.
    status = request["status"]
    # Increment the count for this status at this endpoint.
    status_by_path[path][status] += 1
# Convert the outer defaultdict to an ordinary dictionary.
result = dict(status_by_path)

# Print the status counts for every endpoint.
print(result)
# Print only the counts for the users endpoint.
print(result["/users"])
```
