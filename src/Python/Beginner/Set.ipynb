```python
#Set 
"""A set stores unique elements, 
so repeated values don't create additional entries.""" 

required_skills = {"python", "http", "testing"}
known_skills = {"python", "sql", "testing"}
missing_skills = required_skills - known_skills
shared_skills = required_skills & known_skills
all_skills = required_skills | known_skills
knows_python = "python" in known_skills

print("Missing skills:", sorted(missing_skills))
print("Shared skills:", sorted(shared_skills))
print("All skills:", sorted(all_skills))
print("Knows Python:", knows_python)
```
