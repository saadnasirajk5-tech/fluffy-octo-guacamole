```python
# Create a list containing two tasks.
original = ["draft", "review"]
# Make another variable refer to the same list object.
alias = original
# Modify the shared list through the alias variable.
alias.append("publish")
# Print the original variable to inspect the shared modification.
print("Original after alias append:", original)
# Create a shallow copy of the current list.
copied_tasks = original.copy()
# Append a task to the copied list, not the original list.
copied_tasks.append("archive")
# Print the original list after modifying the copy.
print("Original after copy append:", original)
# Print the separate copy with its additional item.
print("Copied list:", copied_tasks)
# Check whether the original and alias are the same object.
print("Original is alias:", original is alias)
# Check whether the original and copy are the same object.
print("Original is copy:", original is copied_tasks)
```

```python
"""The is operator checks object identity whether two references
point to the same object.  == is equal values.
"""
Saad = ["Hi","Ali"] 
Saad = Ali  
print(Ali)
print(Ali is Saad)
```
