```python
# Store the original usernames in a list.
usernames = [" Saadi ", "ALI", "  sara", "ali "]
cleaned = [] # To hold transformed usernames 
for username in usernames: 
    # Remove surrounding whitespace and convert the name to lowercase
    normalized = username.strip().lower() 
    cleaned.append(normalized) 
print(cleaned)
```

```python
# With comprehension  
# Store the original usernames in a list.
usernames = [" Saadi ", "ALI", "  sara", "ali "] 
cleaned = [username.strip().lower() for username in usernames] 
print(cleaned)  
long_names = [name for name in cleaned if len(name)>3] 
print(long_names)
```

```python
# Dictionary comprehensions 
# Store product names and their original prices in a dictionary.
prices = {"keyboard": 50.0, "mouse": 25.0, "monitor": 200.0}
# Build a new dictionary with prices increased by ten percent.
updated_prices = {name: round(price * 1.10, 2) for name, price in prices.items()}
# Print the new dictionary.
print(updated_prices)
# Build a dictionary containing only products priced above fifty dollars.
expensive_products = {name: price for name, price in prices.items() if price > 50}
# Print the filtered dictionary.
print(expensive_products)
```

```python
# Set Comprehensions 

# Store numbers with some duplicates.
numbers = [1, 2, 2, 3, 3, 3, 4, 5]
print(numbers)
# Build a set containing the square of every number.
squares = {number ** 2 for number in numbers}

# Print the unique squared values.
print(sorted(squares))
# Build a set containing only even numbers from the original list.
even_numbers = {number for number in numbers if number % 2 == 0}
# Print the unique even values in sorted order.
print(sorted(even_numbers))
```
