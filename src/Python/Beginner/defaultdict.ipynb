```python
# defaultdict tells python what to do when you access a missing key 
from collections import defaultdict 
# Store sample tasks as dictionaries.
tasks = [
    {"owner": "Saadi", "title": "Build API"},  # First task.
    {"owner": "Ali", "title": "Write tests"},  # Second task.
    {"owner": "Saadi", "title": "Review code"},  # Third task.
]
tasks_by_owner = defaultdict(list) 
for task in tasks: 
    # Retrieve the owner associated with the current task. 
    owner = task["owner"]        
    tasks_by_owner[owner].append(task["title"])           
result = dict(tasks_by_owner) 
print(result)
print(result["Saadi"])
```
